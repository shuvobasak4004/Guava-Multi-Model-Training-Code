```python
from ultralytics import YOLO

# Load YOLOv12 small model from the Ultralytics model hub
model = YOLO("yolo12s.pt")  # ✅ correct model tag for Ultralytics YOLOv12 small

# Train for 50 epochs
model.train(
    data="/content/data.yaml",
    epochs=50,
    imgsz=640,
    batch=16,
    workers=4,
    project="/content/yolo12_runs",
    name="guaal_50epochs_run"
)
```

```python
# 🔽 Zip and download all YOLOv12 training results in one go
from google.colab import files
import shutil, os

# Paths
run_dir = "/content/yolo12_runs/guaal_50epochs_run"
zip_path = "/content/guaal_50epochs_results.zip"

# Zip everything
if os.path.exists(run_dir):
    shutil.make_archive(zip_path.replace(".zip", ""), 'zip', run_dir)
    print(f"✅ Zipped results at: {zip_path}")
    # Start download
    files.download(zip_path)
else:
    print("⚠️ Training folder not found. Check your run path.")
```
